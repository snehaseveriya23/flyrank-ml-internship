# 03: Model Training and Out-of-Time Validation

Evaluates model candidates against a rigorous Out-of-Time (OOT) holdout split (Weeks 10-12).

```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('model_benchmark_results.json', 'r') as f:
    res = json.load(f)
res_df = pd.DataFrame(res).T
res_df
```

### Precision and Ranking Performance: Out-of-Time Split

```python
res_df[['ROC_AUC', 'PR_AUC', 'Precision@Top10%', 'Precision@Top20%']].plot(kind='bar', figsize=(9, 4.5), colormap='magma')
plt.title('Model Evaluation on Out-of-Time Test Split')
plt.ylabel('Metric Score')
plt.ylim(0, 1.05)
plt.xticks(rotation=15, ha='right')
plt.grid(True, linestyle='--', alpha=0.4)
plt.legend(frameon=True)
plt.show()
```
