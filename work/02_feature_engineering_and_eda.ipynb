# 02: Leakage-Safe Feature Engineering

Constructs sliding 4-week lookback indicators and separate 4-week future forecast targets to prevent temporal lookahead bias.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

feat_df = pd.read_csv('engineered_features_dataset.csv')
print('Feature dimensions:', feat_df.shape)
feat_df.head()
```

### Feature Correlation with Future Decay Target

```python
corrs = feat_df.select_dtypes(include=np.number).corr()['target_opportunity_decay'].sort_values()
plt.figure(figsize=(9, 5))
corrs[:-1].plot(kind='barh', color='#1a73e8')
plt.title('Pearson Correlation with 4-Week Decay Target')
plt.xlabel('Correlation')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
