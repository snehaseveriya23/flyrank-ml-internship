## 🎤 ML-12: 5-Minute Showcase Demo Outline

### Minute 1: The Question & Problem (FlyRank Content Decay)
- **Problem**: Reactive content audits identify traffic loss too late—after high-value URLs drop from top SERP positions.
- **Question**: Can we forecast structural organic search decay 4 weeks early to automate proactive refresh playbooks?

### Minute 2: The Method (Leakage-Safe Modeling)
- **Data**: FlyRank Search Intelligence Warehouse (16 weeks, 1,200 URLs).
- **Signals**: 4-week click momentum, dwell time variance, and power-law SERP expected CTR deviation: $CTR_{expected} = 0.30 / (pos^{0.85})$.
- **Validation**: Strict Out-of-Time (OOT) split (Train <= week 9, Test >= week 10).

### Minute 3: The Key Chart & Honest Result
- **Result**: Gradient Boosted Decision Trees achieve **0.8172 PR-AUC** and **90.0% Precision@Top10%** (vs 0.6311 PR-AUC for baseline rule).
- **Honest Framing**: Observational decision support; directional decay risk rather than causal algorithm claims.

### Minute 4: The Ranked Recommendation Engine
- **3-Tier Playbook**: P0 Emergency Overhaul, P0 Snippet CTR Fix, and P1 Freshness/Internal Links with reason codes and estimated recoverable click yield.
