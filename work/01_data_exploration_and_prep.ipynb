### Objectives
1. Ingest multi-week search performance time-series data.
2. Audit missing values, CTR distributions, and page volatility.
3. Verify public-safe anonymization protocols.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('anonymized_flyrank_search_dataset.csv')
print('Total observations:', len(df))
print('Unique pages:', df['page_id'].nunique())
df.head()
```

### Distribution Profiles

```python
df[['impressions', 'clicks', 'observed_ctr', 'avg_position', 'avg_dwell_sec']].describe().T
```

### Click Trajectory by Archetype Cluster

```python
plt.figure(figsize=(10, 4))
sns.lineplot(data=df, x='week_index', y='clicks', hue='archetype_id', palette='tab10')
plt.title('Weekly Clicks by Performance Cluster Archetype')
plt.xlabel('Week Index')
plt.ylabel('Organic Clicks')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```
